```python
import polars as pl
import networkx as nx
import math

def clean_value(v):
    # Preserve missing values as empty strings
    if v is None:
        return ""

    # Handle NaN
    try:
        if math.isnan(v):
            return ""
    except (TypeError, ValueError):
        pass

    # Convert numpy scalars to native Python types
    if hasattr(v, "item"):
        v = v.item()

    return v


def gen_subgraph(root_ids, neuron_file, connection_file, output_file, subset=None):
    # -----------------------------------------------------------------------------
    # Load data
    # -----------------------------------------------------------------------------

    neurons = pl.read_parquet(neuron_file)
    connections = pl.read_parquet(connection_file)

    root_set = set(root_ids)

    # -----------------------------------------------------------------------------
    # Keep only requested neurons
    # -----------------------------------------------------------------------------

    neurons_sub = neurons.filter(
        pl.col("Root ID").is_in(root_set)
    )

    # -----------------------------------------------------------------------------
    # Keep only edges entirely inside the requested neuron set
    # -----------------------------------------------------------------------------

    connections_sub = connections.filter(
        pl.col("pre_root_id").is_in(root_set)
        & pl.col("post_root_id").is_in(root_set)
    )

    # -----------------------------------------------------------------------------
    # If subset is specified, keep edges which connect TO or FROM the subset
    # -----------------------------------------------------------------------------

    if subset is not None:

        subset_set = set(subset)

        connections_sub = connections_sub.filter(
            pl.col("pre_root_id").is_in(subset_set)
            | pl.col("post_root_id").is_in(subset_set)
        )

    print(f"Nodes: {len(neurons_sub)}")
    print(f"Edges: {len(connections_sub)}")

    # -----------------------------------------------------------------------------
    # Build NetworkX graph
    # -----------------------------------------------------------------------------

    G = nx.DiGraph()

    # Add nodes with every metadata column
    for row in neurons_sub.to_dicts():
        node_id = str(row["Root ID"])

        attrs = {
            k: clean_value(v)
            for k, v in row.items()
            if k != "Root ID"
        }

        G.add_node(node_id, **attrs)

    # Add edges with every metadata column
    for row in connections_sub.to_dicts():

        source = str(row["pre_root_id"])
        target = str(row["post_root_id"])

        attrs = {
            k: clean_value(v)
            for k, v in row.items()
            if k not in ("pre_root_id", "post_root_id")
        }

        G.add_edge(source, target, **attrs)

    # -----------------------------------------------------------------------------
    # Export
    # -----------------------------------------------------------------------------

    nx.write_graphml(G, output_file)

    print(f"Saved {output_file}")

# -----------------------------------------------------------------------------
# User parameters
# -----------------------------------------------------------------------------


NEURON_FILE = "data/banc_neurons_v888.parquet"
CONNECTION_FILE = "data/banc_connections_v888.parquet"

OUTPUT_GRAPHML = "generated/front_right_13a_mn.graphml"
gen_subgraph(id_13a_mns, NEURON_FILE, CONNECTION_FILE, OUTPUT_GRAPHML)

```

```python
import matplotlib.pyplot as plt
target_ids = id_mns
connections = pl.read_parquet(CONNECTION_FILE)
neurons = pl.read_parquet(NEURON_FILE)
# ---------------------------------------------------------
# 1. Get connections INTO the target population
# ---------------------------------------------------------

incoming = connections.filter(
    pl.col("post_root_id").is_in(target_ids)
)

# ---------------------------------------------------------
# 2. Add hemilineage + neurotransmitter information
# ---------------------------------------------------------

incoming = incoming.join(
    neurons.select([
        "Root ID",
        "Hemilineage",
        "Predicted NT type"
    ]),
    left_on="pre_root_id",
    right_on="Root ID",
    how="left"
)
incoming = incoming.with_columns(
    pl.when(pl.col("Hemilineage") == "putative_primary")
    .then(pl.lit("DN"))
    .when(
        pl.col("Hemilineage").is_null() |
        (pl.col("Hemilineage") == "")
    )
    .then(pl.lit("Sensor"))
    .otherwise(pl.col("Hemilineage"))
    .alias("Hemilineage")
)

# ---------------------------------------------------------
# 3. Sum synapses by hemilineage
# ---------------------------------------------------------

hemilineage_synapses = (
    incoming
    .group_by("Hemilineage")
    .agg(
        pl.col("syn_count").sum().alias("synapses")
    )
    .sort("synapses", descending=True)
)

# ---------------------------------------------------------
# 4. Add neurotransmitter for each hemilineage
# ---------------------------------------------------------

hemilineage_nt = (
    incoming
    .group_by("Hemilineage")
    .agg(
        pl.col("Predicted NT type")
        .mode()
        .first()
        .alias("NT")
    )
)

hemilineage_synapses = hemilineage_synapses.join(
    hemilineage_nt,
    on="Hemilineage",
    how="left"
)

# ---------------------------------------------------------
# 5. Calculate percentage of total synaptic input
# ---------------------------------------------------------

total_synapses = hemilineage_synapses["synapses"].sum()

hemilineage_percent = hemilineage_synapses.with_columns(
    (
        pl.col("synapses") / total_synapses * 100
    ).alias("percent")
)

# ---------------------------------------------------------
# 6. Cluster hemilineages contributing <3%
# ---------------------------------------------------------

threshold = 3.0

large = hemilineage_percent.filter(
    pl.col("percent") >= threshold
)

small = hemilineage_percent.filter(
    pl.col("percent") < threshold
)

other_percent = small["percent"].sum()

other = pl.DataFrame({
    "Hemilineage": ["Other (<3%)"],
    "synapses": [small["synapses"].sum()],
    "NT": ["Other"],
    "percent": [other_percent],
})

plot_data = pl.concat([
    large,
    other
])


print(hemilineage_percent)
```

```python
nt_colors = {
    "ACH": "tab:green",
    "GLUT": "tab:orange",
    "GABA": "tab:blue",
    "OCT": "tab:red",
    "DA": "tab:purple",
    "SER": "tab:pink",
}

labels = hemilineage_percent["Hemilineage"].to_list()
sizes = hemilineage_percent["percent"].to_list()
nts = hemilineage_percent["NT"].to_list()

colors = [
    nt_colors.get(nt, "gray")
    for nt in nts
]

plt.figure(figsize=(10, 10))

plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Presynaptic input by hemilineage")
plt.tight_layout()
plt.savefig('preMNs.svg', format='svg')
plt.show()
```

```python
# ---------------------------------------------------------
# 8. Function to create a hemilineage pie chart
# ---------------------------------------------------------

def make_hemilineage_pie(data, title, threshold=3.0):
    # Sum synapses by hemilineage AND NT
    grouped = (
        data
        .group_by(["Hemilineage", "Predicted NT type"])
        .agg(
            pl.col("syn_count").sum().alias("synapses")
        )
    )

    # If a hemilineage has multiple NTs, assign it the NT
    # contributing the most synapses
    grouped = (
        grouped
        .sort("synapses", descending=True)
        .group_by("Hemilineage")
        .agg([
            pl.col("synapses").sum().alias("synapses"),
            pl.col("Predicted NT type").first().alias("NT")
        ])
        .sort("synapses", descending=True)
    )

    # Calculate percentages
    total = grouped["synapses"].sum()

    grouped = grouped.with_columns(
        (pl.col("synapses") / total * 100).alias("percent")
    )

    # Cluster hemilineages < threshold
    large = grouped.filter(
        pl.col("percent") >= threshold
    )

    small = grouped.filter(
        pl.col("percent") < threshold
    )

    other_percent = small["percent"].sum()

    if other_percent > 0:
        other = pl.DataFrame({
            "Hemilineage": ["Other (small)"],
            "synapses": [small["synapses"].sum()],
            "NT": ["Other"],
            "percent": [other_percent],
        })

        plot_data = pl.concat([large, other])
    else:
        plot_data = large

    # -----------------------------------------------------
    # NT colors
    # -----------------------------------------------------

    nt_colors = {
        "ACH": "tab:green",
        "GLUT": "tab:orange",
        "GABA": "tab:blue",
        "OCT": "tab:red",
        "DA": "tab:purple",
        "SER": "tab:pink",
    }


    colors = [
        nt_colors.get(nt, "gray")
        for nt in plot_data["NT"].to_list()
    ]

    # -----------------------------------------------------
    # Plot
    # -----------------------------------------------------

    plt.figure(figsize=(10, 10))

    plt.pie(
        plot_data["percent"].to_list(),
        labels=plot_data["Hemilineage"].to_list(),
        colors=colors,
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title(title)
    plt.tight_layout()
    plt.show()

    return plot_data


# ---------------------------------------------------------
# 9. ACh only
# ---------------------------------------------------------

ach = incoming.filter(
    pl.col("Predicted NT type") == "ACH"
)

ach_plot = make_hemilineage_pie(
    ach,
    "Acetylcholine input by hemilineage",
    threshold=1.0
)


# ---------------------------------------------------------
# 10. Everything except ACh
# ---------------------------------------------------------

non_ach = incoming.filter(
    pl.col("Predicted NT type") != "ACH"
)

non_ach_plot = make_hemilineage_pie(
    non_ach,
    "Non-Acetylcholine input by hemilineage",
    threshold=1.0
)
```

```python
def display_presynaptic_partners(target_ids, connection_file, neuron_file):
    connections = pl.read_parquet(connection_file)
    neurons = pl.read_parquet(neuron_file)
    # ---------------------------------------------------------
    # 1. Get connections INTO the target population
    # ---------------------------------------------------------

    incoming = connections.filter(
        pl.col("post_root_id").is_in(target_ids)
    )

    # ---------------------------------------------------------
    # 2. Add hemilineage + neurotransmitter information
    # ---------------------------------------------------------

    incoming = incoming.join(
        neurons.select([
            "Root ID",
            "Hemilineage",
            "Predicted NT type"
        ]),
        left_on="pre_root_id",
        right_on="Root ID",
        how="left"
    )
    incoming = incoming.with_columns(
        pl.when(pl.col("Hemilineage") == "putative_primary")
        .then(pl.lit("DN"))
        .when(
            pl.col("Hemilineage").is_null() |
            (pl.col("Hemilineage") == "")
        )
        .then(pl.lit("Sensor"))
        .otherwise(pl.col("Hemilineage"))
        .alias("Hemilineage")
    )

    # ---------------------------------------------------------
    # 3. Sum synapses by hemilineage
    # ---------------------------------------------------------

    hemilineage_synapses = (
        incoming
        .group_by("Hemilineage")
        .agg(
            pl.col("syn_count").sum().alias("synapses")
        )
        .sort("synapses", descending=True)
    )

    # ---------------------------------------------------------
    # 4. Add neurotransmitter for each hemilineage
    # ---------------------------------------------------------

    hemilineage_nt = (
        incoming
        .group_by("Hemilineage")
        .agg(
            pl.col("Predicted NT type")
            .mode()
            .first()
            .alias("NT")
        )
    )

    hemilineage_synapses = hemilineage_synapses.join(
        hemilineage_nt,
        on="Hemilineage",
        how="left"
    )

    # ---------------------------------------------------------
    # 5. Calculate percentage of total synaptic input
    # ---------------------------------------------------------

    total_synapses = hemilineage_synapses["synapses"].sum()

    hemilineage_percent = hemilineage_synapses.with_columns(
        (
            pl.col("synapses") / total_synapses * 100
        ).alias("percent")
    )


       


    print(hemilineage_percent)

    # Total
    total_plot = make_hemilineage_pie(
        incoming,
        "Presynaptic input by hemilineage",
        threshold=0.0
    )

    # Excitatory
    ach = incoming.filter(
        pl.col("Predicted NT type") == "ACH"
    )

    ach_plot = make_hemilineage_pie(
        ach,
        "Acetylcholine input by hemilineage",
        threshold=1.0
    )


    #Inhibitory

    non_ach = incoming.filter(
        pl.col("Predicted NT type") != "ACH"
    )

    non_ach_plot = make_hemilineage_pie(
        non_ach,
        "Non-Acetylcholine input by hemilineage",
        threshold=1.0
    )
    return total_plot, ach_plot, non_ach_plot

total_plot, ach_plot, non_ach_plot = display_presynaptic_partners(id_mns, CONNECTION_FILE, NEURON_FILE)
plt.show()
```

```python
def display_postsynaptic_partners(target_ids, connection_file, neuron_file):
    connections = pl.read_parquet(connection_file)
    neurons = pl.read_parquet(neuron_file)

    # ---------------------------------------------------------
    # 1. Get connections OUT OF the target population
    # ---------------------------------------------------------

    outgoing = connections.filter(
        pl.col("pre_root_id").is_in(target_ids)
    )

    # ---------------------------------------------------------
    # 2. Add hemilineage + neurotransmitter of downstream neuron
    # ---------------------------------------------------------

    outgoing = outgoing.join(
        neurons.select([
            "Root ID",
            "Hemilineage",
            "Predicted NT type"
        ]),
        left_on="post_root_id",
        right_on="Root ID",
        how="left"
    )

    outgoing = outgoing.with_columns(
        pl.when(pl.col("Hemilineage") == "putative_primary")
        .then(pl.lit("DN"))
        .when(
            pl.col("Hemilineage").is_null() |
            (pl.col("Hemilineage") == "")
        )
        .then(pl.lit("Sensor"))
        .otherwise(pl.col("Hemilineage"))
        .alias("Hemilineage")
    )

    # ---------------------------------------------------------
    # 3. Sum synapses by postsynaptic hemilineage
    # ---------------------------------------------------------

    hemilineage_synapses = (
        outgoing
        .group_by("Hemilineage")
        .agg(
            pl.col("syn_count").sum().alias("synapses")
        )
        .sort("synapses", descending=True)
    )

    # ---------------------------------------------------------
    # 4. Determine the predominant NT of each downstream
    #    hemilineage, for coloring
    # ---------------------------------------------------------

    hemilineage_nt = (
        outgoing
        .group_by("Hemilineage")
        .agg(
            pl.col("Predicted NT type")
            .mode()
            .first()
            .alias("NT")
        )
    )

    hemilineage_synapses = hemilineage_synapses.join(
        hemilineage_nt,
        on="Hemilineage",
        how="left"
    )

    # ---------------------------------------------------------
    # 5. Calculate percentages
    # ---------------------------------------------------------

    total_synapses = hemilineage_synapses["synapses"].sum()

    hemilineage_percent = hemilineage_synapses.with_columns(
        (
            pl.col("synapses") / total_synapses * 100
        ).alias("percent")
    )

    print(hemilineage_percent)

    # ---------------------------------------------------------
    # 6. Combine hemilineages contributing <3% into "Other"
    # ---------------------------------------------------------

    threshold = 0.0

    large = hemilineage_percent.filter(
        pl.col("percent") >= threshold
    )

    small = hemilineage_percent.filter(
        pl.col("percent") < threshold
    )

    if small.height > 0:
        other = pl.DataFrame({
            "Hemilineage": ["Other (<3%)"],
            "synapses": [small["synapses"].sum()],
            "NT": ["Other"],
            "percent": [small["percent"].sum()],
        })

        plot_data = pl.concat([large, other])
    else:
        plot_data = large

    # ---------------------------------------------------------
    # 7. Colors based on downstream neurotransmitter
    # ---------------------------------------------------------

    nt_colors = {
        "ACH": "tab:blue",
        "GABA": "tab:green",
        "GLUT": "tab:orange",
        "OA": "tab:red",
        "DA": "tab:purple",
        "5HT": "tab:pink",
        "Other": "gray",
    }

    colors = [
        nt_colors.get(nt, "gray")
        for nt in plot_data["NT"].to_list()
    ]

    # ---------------------------------------------------------
    # 8. Plot
    # ---------------------------------------------------------

    plt.figure(figsize=(10, 10))

    plt.pie(
        plot_data["percent"].to_list(),
        labels=plot_data["Hemilineage"].to_list(),
        colors=colors,
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title("Postsynaptic targets by hemilineage")
    plt.tight_layout()

    return plot_data

plot_data = display_postsynaptic_partners(id_cpg_tut_core, CONNECTION_FILE, NEURON_FILE)
plt.savefig('postCPG.svg', format='svg')
```
